Payments — Date Cleaning & Business-Rule Validation

```python


# Convert both date columns to proper datetime format.
# errors='coerce' converts genuinely invalid date values to NaT

payments["Due_Date"] = pd.to_datetime(
    payments["Due_Date"],
    errors="coerce"
)

payments["Payment_Date"] = pd.to_datetime(
    payments["Payment_Date"],
    errors="coerce"
)

# Check whether any invalid/missing dates were created during conversion.

print("Missing Due_Date:", payments["Due_Date"].isna().sum())
print("Missing Payment_Date:", payments["Payment_Date"].isna().sum())
```

Payment_Date is missing for 9,561 records, but these records represent
missed payments. Since no payment occurred, a Payment_Date is not applicable.
Therefore, these NaN values are retained rather than artificially filled.

Business-rule validation

```python

# Rule 1:
# A Paid payment should have a Payment_Date.
paid_without_date = payments[
    (payments["Payment_Status"] == "Paid") &
    (payments["Payment_Date"].isna())
]

# Rule 2:
# A Paid payment should not have a Payment_Date earlier than Due_Date.
paid_before_due = payments[
    (payments["Payment_Status"] == "Paid") &
    (payments["Payment_Date"] < payments["Due_Date"])
]

# Rule 3:
# A Missed payment should not have an actual amount paid.
missed_with_payment = payments[
    (payments["Payment_Status"] == "Missed") &
    (payments["Amount_Paid"] > 0)
]

# Rule 4:
# A Missed payment should not have a Payment_Date.
missed_with_date = payments[
    (payments["Payment_Status"] == "Missed") &
    (payments["Payment_Date"].notna())
]

print("Paid payments without Payment_Date:", len(paid_without_date))
print("Paid payments before Due_Date:", len(paid_before_due))
print("Missed payments with Amount_Paid > 0:", len(missed_with_payment))
print("Missed payments with Payment_Date:", len(missed_with_date))
```

```python
# Keep the raw Payments table unchanged.
# All cleaning operations will be performed on this copy.

payments_clean = payments.copy()

# Convert date fields to proper datetime format.
payments_clean["Due_Date"] = pd.to_datetime(
    payments_clean["Due_Date"],
    errors="coerce"
)

payments_clean["Payment_Date"] = pd.to_datetime(
    payments_clean["Payment_Date"],
    errors="coerce"
)
```

```python
import sqlite3

# Create an in-memory SQL database.
# This allows us to practice SQL directly on our Pandas DataFrames
# without needing to load the data into MySQL first.
conn = sqlite3.connect(":memory:")

# Register the cleaned Loans table as a SQL table.
loans_clean.to_sql(
    "loans",
    conn,
    index=False,
    if_exists="replace"
)

# Business Question:
# Which loan types have the highest loan volume and total loan amount?

query = """
SELECT
    Loan_Type,
    COUNT(Loan_ID) AS Loan_Count,
    ROUND(SUM(Loan_Amount), 2) AS Total_Loan_Amount,
    AVG(Loan_Amount) AS Avg_Loan_Amount
FROM loans
GROUP BY Loan_Type
ORDER BY Total_Loan_Amount DESC;
"""

portfolio = pd.read_sql(query, conn)

portfolio
```

Which loan types have the highest default exposure?

```python
query = """
SELECT
    Loan_Type,
    COUNT(Loan_ID) AS Default_Loan_Count,
    ROUND(SUM(Loan_Amount), 2) AS Default_Amount
FROM loans
WHERE Loan_Status = 'Default'
GROUP BY Loan_Type
ORDER BY Default_Amount DESC;
"""

default_by_type = pd.read_sql(query, conn)

default_by_type
```

```python
query = """
SELECT
    Loan_Type,

    -- Total number of loans for each loan type
    COUNT(Loan_ID) AS Total_Loans,

    -- Number of loans that are in Default
    SUM(CASE
        WHEN Loan_Status = 'Default' THEN 1
        ELSE 0
    END) AS Default_Loans,

    -- Percentage of loans that are in Default
    ROUND(
        100.0 * SUM(CASE
            WHEN Loan_Status = 'Default' THEN 1
            ELSE 0
        END) / COUNT(Loan_ID),
        2
    ) AS Default_Rate,

    -- Total money currently exposed through defaulted loans
    ROUND(
        SUM(CASE
            WHEN Loan_Status = 'Default' THEN Loan_Amount
            ELSE 0
        END),
        2
    ) AS Default_Amount

FROM loans

GROUP BY Loan_Type

ORDER BY Default_Rate DESC;
"""

default_analysis = pd.read_sql(query, conn)

default_analysis
```

Which customer employment groups have the largest loan exposure and default risk?

```python
# Register the cleaned Customers table in the same SQLite connection.
# The SQL JOIN below can then connect customers to loans using Customer_ID.

customers_clean.to_sql(
    "customers",
    conn,
    index=False,
    if_exists="replace"
)

# Now run the customer-segment analysis.
query = """
SELECT
    c.Employment_Type,

    -- Number of loans for each employment group
    COUNT(l.Loan_ID) AS Total_Loans,

    -- Total loan exposure for each employment group
    ROUND(SUM(l.Loan_Amount), 2) AS Total_Loan_Amount,

    -- Number of defaulted loans
    SUM(
        CASE
            WHEN l.Loan_Status = 'Default' THEN 1
            ELSE 0
        END
    ) AS Default_Loans,

    -- Default percentage within each employment group
    ROUND(
        100.0 * SUM(
            CASE
                WHEN l.Loan_Status = 'Default' THEN 1
                ELSE 0
            END
        ) / COUNT(l.Loan_ID),
        2
    ) AS Default_Rate

FROM loans l

JOIN customers c
    ON l.Customer_ID = c.Customer_ID

GROUP BY c.Employment_Type

ORDER BY Total_Loan_Amount DESC;
"""

employment_analysis = pd.read_sql(query, conn)

employment_analysis
```

```python
# Register the cleaned Branches table so SQL can JOIN it with Loans.
branches_clean.to_sql(
    "branches",
    conn,
    index=False,
    if_exists="replace"
)

query = """
SELECT
    b.Branch_ID,
    b.Branch_Name,
    b.City,
    b.Region,

    -- Number of loans handled by the branch
    COUNT(l.Loan_ID) AS Loan_Count,

    -- Total amount of loans handled by the branch
    ROUND(SUM(l.Loan_Amount), 2) AS Total_Loan_Amount,

    -- Average loan size at the branch
    ROUND(AVG(l.Loan_Amount), 2) AS Avg_Loan_Amount

FROM loans l

JOIN branches b
    ON l.Branch_ID = b.Branch_ID

GROUP BY
    b.Branch_ID,
    b.Branch_Name,
    b.City,
    b.Region

ORDER BY Total_Loan_Amount DESC;
"""

branch_analysis = pd.read_sql(query, conn)

branch_analysis
```

How frequently are payments late, and which loan types have the highest late-payment rate?

```python
# Register the cleaned Payments table in our SQL connection.
# This makes the current payments DataFrame available to SQL.
payments_clean = payments.copy()

payments_clean.to_sql(
    "payments",
    conn,
    index=False,
    if_exists="replace"
)

# Analyze repayment performance by loan type.
query = """
SELECT
    l.Loan_Type,

    -- Number of payment records
    COUNT(p.Payment_ID) AS Total_Payments,

    -- Number of payments that were late
    SUM(
        CASE
            WHEN p.Days_Late > 0 THEN 1
            ELSE 0
        END
    ) AS Late_Payments,

    -- Percentage of payments that were late
    ROUND(
        100.0 * SUM(
            CASE
                WHEN p.Days_Late > 0 THEN 1
                ELSE 0
            END
        ) / COUNT(p.Payment_ID),
        2
    ) AS Late_Payment_Rate,

    -- Average delay among late payments
    ROUND(
        AVG(
            CASE
                WHEN p.Days_Late > 0 THEN p.Days_Late
            END
        ),
        2
    ) AS Avg_Days_Late

FROM payments p

JOIN loans l
    ON p.Loan_ID = l.Loan_ID

GROUP BY l.Loan_Type

ORDER BY Late_Payment_Rate DESC;
"""

payment_analysis = pd.read_sql(query, conn)

payment_analysis
```

```python
# Calculate loan processing time directly from the cleaned DataFrames.
# This avoids depending on the SQL copy of the tables.

processing = loan_applications_clean[
    ["Application_ID", "Loan_Type", "Application_Date", "Decision_Date"]
].copy()

# Calculate the number of days taken to make the application decision.
processing["Decision_Days"] = (
    processing["Decision_Date"] - processing["Application_Date"]
).dt.total_seconds() / (24 * 60 * 60)

# Keep only applications that have a corresponding loan.
processing = processing.merge(
    loans_clean[
        ["Application_ID", "Approval_Date", "Disbursement_Date"]
    ],
    on="Application_ID",
    how="inner"
)

# Calculate the number of days between approval and disbursement.
processing["Disbursement_Days"] = (
    processing["Disbursement_Date"] - processing["Approval_Date"]
).dt.total_seconds() / (24 * 60 * 60)

# Standardize Loan_Type before grouping.
# This ensures values such as "personal loan" and "Personal Loan"

processing["Loan_Type"] = (
    processing["Loan_Type"]
    .str.strip()
    .str.title()
)

# Calculate the final business metrics.
processing_analysis = (
    processing
    .groupby("Loan_Type")
    .agg(
        Avg_Decision_Days=("Decision_Days", "mean"),
        Avg_Disbursement_Days=("Disbursement_Days", "mean")
    )
    .round(2)
    .sort_values("Avg_Decision_Days", ascending=False)
)

processing_analysis
```

```python
# Business Question :
# What is the application approval rate by loan type?

query = """
SELECT
    Loan_Type,

    -- Total number of applications received
    COUNT(Application_ID) AS Total_Applications,

    -- Number of approved applications
    SUM(
        CASE
            WHEN Application_Status = 'Approved' THEN 1
            ELSE 0
        END
    ) AS Approved_Applications,

    -- Number of rejected applications
    SUM(
        CASE
            WHEN Application_Status = 'Rejected' THEN 1
            ELSE 0
        END
    ) AS Rejected_Applications,

    -- Percentage of applications that were approved
    ROUND(
        100.0 * SUM(
            CASE
                WHEN Application_Status = 'Approved' THEN 1
                ELSE 0
            END
        ) / COUNT(Application_ID),
        2
    ) AS Approval_Rate

FROM loan_applications

GROUP BY Loan_Type

ORDER BY Approval_Rate DESC;
"""

application_analysis = pd.read_sql(query, conn)

application_analysis
```
